```python
# Problem 1: Mobile Recharge Validation System

def check_recharge(price, plan):
    valid_plans = [199,299,399,599]
    if price < 50:
        print("Recharge Rejected! Price should be minimum 50")
        return False
    if plan not in valid_plans:
        print("Invalid Plan")
        return False
    return True

while True:
    n=int(input("Enter Price: "))
    p=int(input("Enter Plan:"))
    result = check_recharge(n,p)
    if result:
        print("Recharge Successful")
        break
    else:
        print("Try Again\n")
```

```python
# Problem 2: Inventory Reorder Alert System

def check_inventory(products):
    for product,stock in products.items():
        if stock < 15:
            print(product, "-> Reorder Alert")
        else:
            print(product, "-> Stock OK")
store_items={}
n = int(input("Enter no. of products:"))
for i in range(n):
    name = input("Enter product name:")
    quantity = int(input("Enter stock quantity: "))
    store_items[name] = quantity
print("\n Inventatory Status:")
check_inventory(store_items)
    
```

```python
# Problem 3: Student Result Processing System

def process_result(marks):
    total = 0
    for mark in marks:
        total += mark
    avg = total / len(marks)

    if avg >= 50:
        return avg, "Pass"
    else:
        return avg, "Fail"

#User input
n = int(input("Enter number of subjects: "))
marks_list = []
for i in range(n):
    mark = int(input("Enter marks: "))
    marks_list.append(mark)
avg,result = process_result(marks_list)

print("Average Marks: ", avg)
print("Result : ", result)
```

```python
# Problem 4: Cab Fare Estimator with Retry Option

def fare_price(dist,peak):
    base_fare = 50
    per_km_rate = 12
    fare = base_fare + (dist * per_km_rate)
    if peak.lower() == "yes":
        fare = fare + (0.25 * fare) # 25% extra
    return fare

while True:
    dist = float(input("Enter Distance in km: "))
    peak_hour = input("Is it peak hour?(yes/no):")
    total_fare = fare_price(dist, peak_hour)
    print("Total Fare: rs", total_fare)
    retry = input("Do you want to calculate again? (yes/no): ")
    if retry.lower() != "yes":
        print("Thank you for using Cab Estimator")
        break
```

```python
# Problem 5: Employee Attendance Eligibility Checker

def check_eligibility(list):
    cnt = 0
    for status in list:
        if status == "P":
            cnt += 1
    percentage = (cnt / len(list)) * 100
    if percentage >= 75:
        return percentage, "Eligible"
    else:
        return percentage, "Not Eligible"

n = int(input("Enter no. of working days: "))
attendance = []
for i in range(n):
    day = input("Enter attendance(P/A):")
    attendance.append(day.upper())
percent, result = check_eligibility(attendance)
print("Attendance Percentage:", percent)
print("Status :", result)
    
```

```python
# Problem 6: Password Strength Checker

def check_password(password):
    num = False
    special = False
    for i in password:
        if i.isdigit():
            num = True
        if i=="@" or i=="#" or i=="$":
            special = True
    if len(password) >= 8 and num and special:
        return "Strong Password"
    else:
        return "Weak Password"

p=input("Enter password:")
res = check_password(p)
print(res)
```
