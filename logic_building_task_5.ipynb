```python
# Problem 1:Social Media-Post Engagement Analyzer

def social_media_analyzer(likes):
    total_likes = 0
    for like in likes:
        total_likes += like
        
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    return status
#user input
num = int(input("Enter number of posts: "))
likes = []
for i in range(num):
    like = int(input("Enter likes for post:"))
    likes.append(like)
result = social_media_analyzer(likes)
print("Status:",result)
```

```python
# Problem 3: Agriculture-Rainfall Adequacy Checker

def check_rainfall(lists):
    total = 0
    for list in lists:
        total = total + list

    avg = total / len(lists)
    if avg >= 70:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    return avg,status
#User input
num = int(input("Enter number of months:"))
lists = []
for i in range(num):
    list = int(input("Enter rainfall for month: "))
    lists.append(list)
avg,result = check_rainfall(lists)
print("Average Rainfall:",int(avg))
print("Rainfall Status:",result)
```

```python
# Problem 4: Social Media-Duplicate Account Detection

def check_duplicate_accounts(user_list):
    users = set(user_list)

    if len(users) != len(user_list):
        return "Yes"
    else:
        return "No"
#user input
num = int(input("Enter number of usernames:"))
user_list = []
for i in range(num):
    name = input("Enter username:")
    user_list.append(name)
result = check_duplicate_accounts(user_list)
print("Duplicate Accounts Found:",result)
```

```python
# Problem 6: Agriculture-Premium Crop Price Filter

def filter_premium_crops(price_list):
    list = []
    for price in price_list:
        if price > 2000:
            list.append(price)
    return list
num = int(input("Enter number of crops:"))
price_list = []
for i in range(num):
    price = int(input("Enter crop price:"))
    price_list.append(price)
result = filter_premium_crops(price_list)
print("Premium Crops:",result)
```

```python
# Problem 7: System Monitoring-Application Health Checker

def monitoring_health(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    print("Error Count:",errors)
    print("System Status:",status)

#user input
errors = int(input("Enter error count: "))
monitoring_health(errors)
```

```python
# Problem 9: E-Learning - Student Attendance Eligibility

def check_StudentAttendanceEligibility(attendance_list):
    total_classes = len(attendance_list)
    cnt = 0
    for mark in attendance_list:
        if mark == 1:
            cnt+=1
    percentage =(cnt / total_classes) * 100
    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not eligible"
    print("Attendance percantage:", percentage)
    print("Exam Eligibility:",status)

#User input
n = int(input("Enter total number of classes: "))
attendance_list = []
for i in range(n):
    mark = int(input("Enter 1 for Present, 0 for Absent: "))
    attendance_list.append(mark)
check_StudentAttendanceEligibility(attendance_list)
```

```python
# Problem 10: Smart Electricity Bill Analyzer

def calculate_bill(units):
    bill = 0
    if units <= 100:
        bill = units*3
    elif units <= 200:
        bill = (100*3) + ((units - 100)*5)
    else:
        bill = (100*3)+(100*5)+((units - 200)*7)
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"
    return bill,status
units = int(input("Enter total units consumed:"))
total_bill,status = calculate_bill(units)
print("Total Bill:",total_bill)
print("Usage Classification:",status)
```
