```python
# Problem 1: Smart Light Controller

class SmartLight:
    def __init__(self,name):
        self.name = name
        self.status = "OFF"
    def light_on(self):
        self.status = "ON"
        print("Light turned ON")
    def light_off(self):
        self.status = "OFF"
        print("Light turned OFF")
    def show_light(self):
        print(self.name, "is" ,self.status)
#Create object
light = SmartLight("Bedroom Light")
light.light_on()
light.show_light()
        
```

```python
# Problem 2: Employee ID Card System

class EmployeeId:
    def __init__(self, name, id, department):
        self.name = name
        self.id = id
        self.department = department
    def show_idDetails(self):
        print("Employee ID Card")
        print("Name:",self.name)
        print("ID:",self.id)
        print("Department:",self.department)
#Create object
emp = EmployeeId("Rahul","EMP102","IT")
emp.show_idDetails()
```

```python
# Problem 3: Mobile Contact Record

class MobileContact:
    def __init__(self, name, ph_number):
        self.name = name
        self.ph_number = ph_number
    def show_contactDetails(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.ph_number)
#Create object
contact = MobileContact("Anita","9876543210")
contact.show_contactDetails()
```

```python
# Problem 4: Product Price Tag Generator

class Product_price:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def show_priceTag(self):
        print("Product:",self.name)
        print(f"Price: ₹{self.price}")
#Create object
price = Product_price("Headphones", 2499)
price.show_priceTag()
```

```python
# Problem 5: Movie Rating Display System

class MovieRating:
    def __init__(self,name,rating):
        self.name = name
        self.rating = rating
    def display_details(self):
        print("Movie:",self.name)
        print(f"Rating: {self.rating} / 5")
#Create object
rating = MovieRating("Inception",4.8)
rating.display_details()
```
