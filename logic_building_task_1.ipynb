```python
## 1 : User Login Check

user = input("Enter username: ")
pwd = input("Enter Password: ")

username = "admin"
password = "1234"

if (user == username and pwd == password):
    print("Login Successful")
else:
    print("Invalid Credentials")
    
```

```python
## 2 : Pass / Fail Analyzer

marks=[45,78,90,33,60]
pass_cnt = 0
fail_cnt = 0
for mark in marks:
    if mark >= 50:
        pass_cnt += 1
    else:
        fail_cnt += 1
print("Total Pass Students: ",pass_cnt)
print("Total Fail Students: ",fail_cnt)
```

```python
## 3 : Simple Data Cleaner

names = ["Alice", "bob", "CHARLIE"]
cleaned_names = []

for name in names:
    name = name.strip()
    name = name.lower()
    cleaned_names.append(name)
print("Cleaned List:", cleaned_names)
```

```python
## 4 : Message Length Analyzer

texts = ["Hi", "Welcome to the platform", "OK"]
for msg in texts:
    length = len(msg)
    print("Message: ", msg)
    print("Length: ",length)

    if length > 10:
        print("This message is longer than 10 characters")
        print()
```

```python
## 5 :  Error Message Detector

logs = ["INFO","ERROR", "WARNING", "ERROR"]
err_cnt = 0

for log in logs:
    if log == "ERROR":
        err_cnt += 1
print("Total ERROR entries: ", err_cnt)
```
