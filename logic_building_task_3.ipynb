```python
# 1: Employee Performance Bonus Eligibility

employees = {
    "Ravi":92,
    "Anita":88,
    "Kiran":92,
    "Suresh":85
}
high_score = 0
for score in employees.values():
    if score > high_score:
        high_score = score

top_performers = []
for name,score in employees.items():
    if score == high_score:
        top_performers.append(name)

print("Top Performers Eligible for Bonus:", end=" ")
for i in range(len(top_performers)):
    if i == len(top_performers) - 1:
        print(top_performers[i], end="")
    else:
        print(top_performers[i] + ", ",end="")
print("(Score:", high_score,")")
```

```python
# 2: Search Query Keyword Analysis

query = "Buy mobile phone buy phone online"
query = query.lower()
query = query.replace(".","")
query = query.replace(",","")
query = query.replace("!","")
query = query.replace("?","")

words = query.split()
word_cnt = {}

for word in words:
    if word in word_cnt:
        word_cnt[word] += 1
    else:
        word_cnt[word] = 1

res = {}
for word, cnt in word_cnt.items():
    if cnt > 1:
        res[word]= cnt
print(res)
```

```python
# 3: Sensor Data Validation

sensor_readings = [3,4,7,8,10,12,5]
valid_readings = []

for index in range(len(sensor_readings)):
    val = sensor_readings[index]

    if val % 2 == 0:
        valid_readings.append((index,val))
print("Valid Sensor Readings (Hour, Value):", valid_readings)
```

```python
# 4: Email Domain Usage Analysis

emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
cnt = {}
for email in emails:
    domain = email.split("@")[1]
    if domain in cnt:
        cnt[domain] += 1
    else:
        cnt[domain] = 1
total_emails = len(emails)
for domain, count in cnt.items():
    percentage = (count / total_emails)*100
    print(domain + ":", int(percentage), "%")
```

```python
# 5: Sales Spike Detection

sales = [1200, 1500, 900, 2200, 1400, 3000]
total = 0

for val in sales:
    total += val
avg = total / len(sales)
threshold = avg * 1.2

for index in range(len(sales)):
    if sales[index] > threshold:
        print("Day ",index+1,": ",sales[index],sep="")
```
