```python
# Problem 1: Smart Parking lot Management System

def parking_system(capacity, logs):
    curr_cnt = 0
    for log in logs:
        if log == "IN":
            curr_cnt += 1
        elif log == "OUT":
            if curr_cnt > 0:
                curr_cnt -= 1
    print("Currently parked vehicles:",curr_cnt)
    if curr_cnt > capacity:
        print("Parking Status: Full")
    else:
        print("Parking Status: Available")
capacity= int(input("Enter parking capacity:"))
num = int(input("Enter number of vehicle logs:"))
logs = []
for i in range(num):
    entry = input("Enter log(IN/OUT):")
    logs.append(entry)
parking_system(capacity,logs)
```

```python
# Problem 2: Online Food Delivery Time estimator

def delivery_time_estimator(dist, traffic,weather):
    time = dist*5
    if traffic == "Low":
        time += 5
    elif traffic == "Medium":
        time += 10
    elif traffic == "High":
        time += 15
    if weather == "Rainy":
        time += 10
    elif weather == "Stormy":
        time += 20
    return time
dist = int(input("Enter distance(km): "))
traffic = input("Enter traffic level (Low/Medium/High):")
weather = input("Enter weather conditions(Clear/Rainy/Stormy):")
final_time = delivery_time_estimator(dist,traffic,weather)
print("Estimated Delivery Time:",final_time,"minutes")
```

```python
# Problem 3: Movie Theatre Seat Occupancy

def seat_occupancy(total_seats, booked_list):
    cnt = 0
    for seat in booked_list:
        if seat == 1:
            cnt += 1
        occupancy = (cnt/total_seats)*100
    if occupancy == 100:
        status = "Houseful"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"
    print("Occupancy:", int(occupancy),"%")
    print("Show Status:", status)
    if occupancy >= 90:
        print("Suggestion: Open additional show")
    else:
        print("Suggestion: No additional show")

total_seats = int(input("Enter total seats:"))
num = int(input("Enter number of booked seats:"))
booked_list = []
for i in range(num):
    seat = int(input("Enter 1 for booked seat:"))
    booked_list.append(seat)
seat_occupancy(total_seats, booked_list)
```

```python
# Problem 4: Cloud Server Load Classification System

def classify_server_load(cpu_list):
    total = 0
    for value in cpu_list:
        total += value
    avg = total / len(cpu_list)
    if avg < 50:
        status = "Normal"
    elif avg <= 50:
        status = "Warning"
    else:
        status = "Critical"
    print("Average CPU Load:",int(avg),"%")
    print("Server Status:",status)

num = int(input("Enter number of CPU readings:"))
cpu_list = []
for i in range(num):
    reading = int(input("Enter CPU usage(%):"))
    cpu_list.append(reading)
classify_server_load(cpu_list)
```

```python
# Problem 5: Smart Classroom Resource Usage Monitor

def monitor_usage(resources):
    overused = []
    for name in resources:
        if resources[name] > 8:
            overused.append(name)
    if len(overused) > 0:
        alert = "Yes"
    else:
        alert = "No"
    print("Overused Resouces:", ",".join(overused) if overused else "None")
    print("Energy Alert:",alert)

num = int(input("Enter number of resources:"))
resources = {}
for i in range(num):
    name = input("Enter resource name:")
    hours = int(input("Enter usage hours:"))
    resources[name] = hours
monitor_usage(resources)
```

```python
# Problem 6: Online Event Registration Capacity Controller

def capacity_controller(capacity,registrations):
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    print("Confirmed Registrations:",confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

capacity = int(input("Enter event capacity:"))
registrations = int(input("Enter total registrations:"))
capacity_controller(capacity,registrations)
```
