```python
#Program 1: Unique Words in a Sentence

text = input("Enter Sentrnce:")
word_list = text.split()
unique_list = []
for word in word_list:
    if word not in unique_list:
        unique_list.append(word)

print("Unique words count: ",len(unique_list))
print("Unique words :",tuple(unique_list))
```

```python
#Problem 2: Highest Salary from Employee Data

employees = {
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}

highest_salary = 0
highest_employee = ""

for name in employees:
    if employees[name] > highest_salary:
        highest_salary = employees[name]
        highest_employee = name

print("Highest Salary", highest_employee,"-", highest_salary)
```

```python
#Problem 3: Find Maximum and Minimum Values

nums = [45,22,89,10,66]
max = nums[0]
min = nums[0]

for num in nums:
    if num > max:
        max = num
    if num < min:
        min = num

print("List: ", nums)
print("Max: ", max)
print("Min: ",min)
```

```python
#Problem 4: Count Products Above a Price Threshold

prices = [450,1200,899,1500,300]
cnt = 0

for price in prices:
    if price > 1000:
        cnt += 1
print("Products above 1000:", cnt)
```

```python
#Problem 5: Calculate Attendance Percentage

data = input("Enter attendance separated by space (P/A): ")

attendance = data.split()
cnt_present = 0

for val in attendance:
    if val.upper() == "P":
        cnt_present += 1
total = len(attendance)
result = (cnt_present/total) * 100
print("Attendance percentage: ", result)
```

```python
#Problem 6: Remove Duplicate Phone Numbers

data = input("Enter Phone numbers : ")
nums = data.split()
unique_nums = set(nums)
print("Unique phone numbers: ", tuple(unique_nums))
```

```python
#Problem 7: Count Character Frequency

text = input("Enter a string:")
freq = {}
for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
print("Character Frequency: ", freq)
             
```

```python
#Problem 8: Convert List to Tuple

nums = [10,20,30]
res = tuple(nums)
print("Tuple:", res)
```

```python
#Problem 9: Check if a Key Exists in a Dictionary

employees = {
    "Ravi" : 75000,
    "Anita":68000,
    "Kiran":72000
}
name = input("Enter employees name to check : ")
if name in employees.keys():
    print("Employee exists")
else:
    print("Employee does not exists")
```

```python
#Problem 10: Calculate Average Marks

nums = input("Enter marks : ")
marks = nums.split()
total = 0
for mark in marks:
    total += int(mark)
avg = total / len(marks)
print("Average Marks:", avg)
```
